```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = 'Registro_Naconal_de_Turismo_-_Jamundi_-_Valle_del_Cauca_-_Alcaldia_de_Jamundí_20260518.csv'
df = pd.read_csv(file_path, sep=';')

categoria_counts = df['CATEGORIA'].value_counts()

plt.figure(figsize=(10, 6))
categoria_counts.plot(kind='barh', color='skyblue', edgecolor='black')

plt.title('Distribución de Establecimientos por Categoría Turística en Jamundí', fontsize=14, fontweight='bold')
plt.xlabel('Cantidad de Establecimientos', fontsize=12)
plt.ylabel('Categoría', fontsize=12)
plt.gca().invert_yaxis() 
plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = 'Vista_-_Sucre_-_Centros_Digitales_20260518.csv'
df = pd.read_csv(file_path, sep=';')

estado_counts = df['ESTADO'].value_counts()

plt.figure(figsize=(8, 5))
colors = ['#4CAF50', '#FFC107', '#2196F3', '#F44336'] 
estado_counts.plot(kind='bar', color=colors[:len(estado_counts)], edgecolor='black')

plt.title('Estado de los Centros Digitales en Sucre', fontsize=14, fontweight='bold')
plt.xlabel('Estado del Proyecto', fontsize=12)
plt.ylabel('Cantidad de Centros', fontsize=12)
plt.xticks(rotation=45, ha='right') 
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = 'Accidentalidad_Vial_Guadalajara_De_Buga_20260518.csv'
df = pd.read_csv(file_path, sep=';')

clase_counts = df['CLASE_ACCIDENTE'].value_counts().head(5)

plt.figure(figsize=(8, 8))
plt.pie(
    clase_counts, 
    labels=clase_counts.index, 
    autopct='%1.1f%%',
    startangle=140, 
    colors=['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0']
)

plt.title('Top Clases de Accidentes Viales en Guadalajara de Buga', fontsize=14, fontweight='bold')
plt.tight_layout()

plt.show()
```
